```python
"""
Program Name : Debugging Maximum Element Search
Course       : Design and Analysis of Algorithms (DAA)
Question     : Q39 - Debugging Maximum Element Search Code
Objective    : Find the maximum element in a list by handling
               empty-list edge cases using defensive programming.
"""


def find_max(arr):
    """
    Returns the maximum element in the given list.

    Parameters:
        arr (list): A list of numeric values.

    Returns:
        int/float: Maximum value in the list.

    Raises:
        ValueError: If the input list is empty.
    """

    # Check whether the list is empty
    if len(arr) == 0:
        raise ValueError("Cannot find the maximum of an empty list.")

    # Initialize maximum safely
    max_val = arr[0]

    # Traverse the remaining elements
    for value in arr[1:]:

        # Update maximum when a larger element is found
        if value > max_val:
            max_val = value

    return max_val


# ---------------------- Driver Program ----------------------

print("=" * 60)
print("        Maximum Element Search Program")
print("=" * 60)

try:
    # Input number of elements
    n = int(input("Enter the number of elements: "))

    # Input array elements
    numbers = []

    for i in range(n):
        value = int(input(f"Enter element {i + 1}: "))
        numbers.append(value)

    # Find maximum element
    maximum = find_max(numbers)

    # Display output
    print("\nInput List        :", numbers)
    print("Maximum Element   :", maximum)

except ValueError as error:
    print("\nError:", error)

except Exception as error:
    print("\nUnexpected Error:", error)

# ---------------------- Complexity Analysis ----------------------

print("\n" + "=" * 60)
print("Complexity Analysis")
print("=" * 60)

print("Best Case Time Complexity    : O(n)")
print("Average Case Time Complexity : O(n)")
print("Worst Case Time Complexity   : O(n)")
print("Space Complexity             : O(1)")

print("\nExplanation:")
print("- The algorithm scans each element exactly once to find the maximum.")
print("- No nested loops are used, so the running time is linear.")
print("- Only one extra variable (max_val) is used, so the extra space required is constant.")
```
