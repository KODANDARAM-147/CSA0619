```python
# ---------------------------------------------------------
# Design and Analysis of Algorithms
# Question 39
# Binary Search on Sorted Financial Transaction Data
# ---------------------------------------------------------

def binary_search(transactions, target):

    low = 0
    high = len(transactions) - 1

    while low <= high:

        mid = (low + high) // 2

        if transactions[mid] == target:
            return mid

        elif transactions[mid] < target:
            low = mid + 1

        else:
            high = mid - 1

    return -1


# ---------------- Input ----------------

transactions = []

n = int(input("Enter number of transactions: "))

print("\nEnter transaction IDs in Sorted Order:")

for i in range(n):
    transactions.append(int(input()))

target = int(input("\nEnter Transaction ID to search: "))


# ---------------- Search ----------------

position = binary_search(transactions, target)


# ---------------- Output ----------------

print("\nSorted Transaction List:")
print(transactions)

if position != -1:
    print("\nTransaction Found")
    print("Transaction ID :", target)
    print("Index :", position)
else:
    print("\nTransaction Not Found")


# ---------------- Complexity ----------------

print("\n----------- Complexity Analysis -----------")

print("Input Time Complexity       : O(n)")
print("Input Space Complexity      : O(n)")

print("Binary Search Best Case     : O(1)")
print("Binary Search Average Case  : O(log n)")
print("Binary Search Worst Case    : O(log n)")
print("Binary Search Space         : O(1)")

print("Output Time Complexity      : O(n)")
print("Output Space Complexity     : O(1)")

print("Overall Time Complexity     : O(n)")
print("Overall Space Complexity    : O(n)")
```
