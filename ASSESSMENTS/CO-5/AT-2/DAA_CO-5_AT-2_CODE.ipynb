```python
# Q39. Combination Generation of Size K
# Backtracking Algorithm

def generate_combinations(A, k):
    n = len(A)

    # Handle invalid k
    if k < 0 or k > n:
        print("No valid combinations")
        return

    current = []

    def backtrack(start, remaining):

        # If required elements are selected
        if remaining == 0:
            print(current)
            return

        # Pruning condition
        if n - start < remaining:
            return

        # Try every feasible element
        for i in range(start, n - remaining + 1):

            # Select element
            current.append(A[i])

            # Recursive call
            backtrack(i + 1, remaining - 1)

            # Backtrack
            current.pop()

    # Start backtracking
    backtrack(0, k)


# Main Program
A = [1, 2, 3, 4]
k = 2

print("Combinations of size", k, ":")
generate_combinations(A, k)
```
